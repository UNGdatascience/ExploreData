## <span style="font-weight:bold; font-size:1.1em">The describe() Function</span>

The code below creates the **describe()** funcction which provides a summary of the most important decriptive statistics for a data array.

1. For use with an array:<center><span style="color:blue; font-weight:bold; font-size:1.25em">describe(array_name)</span></center>
2. For use with a table and numeric data column:<center><span style="color:blue; font-weight:bold; font-size:1.25em">describe(table_name, column_name)</span></center>

```python
def describe(data, var = None):
  if isinstance(data, Table):
    data = data.column(var)
  names = np.array(["mean", "std", "n", "min", "Q1", "med", "Q3", "max", "IQR"])
  values = np.array([round(np.mean(data),1), round(stats.tstd(data),2), len(data) - 1,
                     min(data), np.percentile(data, 25), np.median(data), np.percentile(data, 75), 
                     max(data), stats.iqr(data)])
  tab = Table().with_columns('Names', names, 'Values', values)
  transposed_tab = Table().with_columns( zip(tab.column('Names'), tab.column('Values')) ).show()
  return
```

```python
bw_array = births.column('Birth Weight')
describe(bw_array)
```

```python
describe(births, 'Birth Weight')
```

```python
narc = pers.column('Narc')
describe(narc)
```

```python
perf = pers.column('Perf')
describe(perf)
```

## <span style="font-weight:bold; font-size:1.1em">The describes() Function</span>

The code below creates the **describes()** function which provides important decriptive statistics with a row for each level of a category variable.<center><span style="color:blue; font-weight:bold; font-size:1.25em">describes(table_name, num_column, group_column)</span></center>

```python
def describes(data, num, group):
  dat = data.column(num)
  group_var = data.column(group)
  groups = np.unique(group_var)
  rows = []
  tab = Table(["group", "mean", "std", "n", "min", "Q1", "median", "Q3", "max", "IQR"])
  for g in groups:
    # Filter data for the current group
    sub_dat = dat[group_var == g]
    rows = [g, round(np.mean(sub_dat), 1), round(stats.tstd(sub_dat), 2), len(sub_dat) - 1,
            min(sub_dat), np.percentile(sub_dat, 25), np.median(sub_dat), np.percentile(sub_dat, 75),
            max(sub_dat), stats.iqr(sub_dat)]        
    tab = tab.with_row(rows)
  tab.show()
  return 
```

```python
describes(pers, 'Narc' , 'PHS')
```

```python
describes(births, "Birth Weight", "Maternal Smoker")
```

```python
describes(pers, 'Sleep' , 'Yr')
```
